```python
roi = pygaul.Items(name="Henan sheng").geometry()  #获取河南行政区
henan = roi

START_YEAR, END_YEAR = 2015, 2024 #时间范围
REDUCE_SCALE = 1500  #防超限降低空间分辨率， 1.5km

#季节定义
SEASONS = {'spring': [3, 4, 5],'summer': [6, 7, 8],'autumn': [9, 10, 11],'winter': [12, 1, 2]}

#显示中文，保证图表中中文和负号显示正常
plt.rcParams['font.sans-serif'] = ['SimHei']  #用来正常显示中文标签
plt.rcParams['axes.unicode_minus'] = False    #用来正常显示负号
```

```python
# 影像准备与预处理 
def apply_scale_factors(img): # DN值转反射率
    optical = img.select('SR_B.').multiply(0.0000275).add(-0.2) # Reflectance = DN * 0.0000275 - 0.2
    return img.addBands(optical, overwrite=True) #覆盖原波段

def mask_clouds(img): #云掩膜
    qa = img.select('QA_PIXEL')
    
    cloud = qa.bitwiseAnd(1 << 4).eq(0)
    shadow = qa.bitwiseAnd(1 << 2).eq(0)
    snow = qa.bitwiseAnd(1 << 3).eq(0)
    
    mask = cloud.And(shadow).And(snow) #三者均为“真”时，像元有效，非云非阴影非雪
    
     
    return img.updateMask(mask).copyProperties(img, ['system:time_start'])

# Landsat band标准化，同意Landsat7、8波段名
LC8_BANDS = ['SR_B2','SR_B3','SR_B4','SR_B5','SR_B6','SR_B7','QA_PIXEL']
LC7_BANDS = ['SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7','QA_PIXEL']
STD_NAMES = ['blue','green','red','nir','swir1','swir2','QA_PIXEL']

# 合并 L7 + L8
def get_landsat_collection(year): # 获取年影像集合
    
    start_date = f'{year}-01-01'  
    end_date = f'{year+1}-01-01'    
# 获取 Landsat8、7影像    
    l8 = (ee.ImageCollection('LANDSAT/LC08/C02/T1_L2') 
          .filterBounds(roi) #限制河南
          .filterDate(start_date, end_date)#时间
          .filter(ee.Filter.lt('CLOUD_COVER', 80))  # 放宽云量限制
          .map(apply_scale_factors)  #应用反射率比例因子校正
          .select(LC8_BANDS, STD_NAMES) # 波段重命名为标准名称
          .map(mask_clouds)) # 应用云掩膜

    
    l7 = (ee.ImageCollection('LANDSAT/LE07/C02/T1_L2')
          .filterBounds(roi)
          .filterDate(start_date, end_date)
          .filter(ee.Filter.lt('CLOUD_COVER', 80))
          .map(apply_scale_factors)
          .select(LC7_BANDS, STD_NAMES)
          .map(mask_clouds))
    
    return l8.merge(l7) #合并Landsat7、8
```

```python
# 计算 NDVI和 FVC 
def add_ndvi_fvc(imgcol):
    # 首先检查集合是否为空
    size = imgcol.size().getInfo()
    if size == 0:
        return imgcol
    
    # 计算NDVI，NDVI = (NIR - RED) / (NIR + RED)，返回包含 NDVI 的新影像集合 ndvi_col
    ndvi_col = imgcol.map(lambda img: img.addBands(
        img.normalizedDifference(['nir','red']).rename('NDVI')))# 将结果以 “NDVI” 添加为新波段。
    
    # 使用分位数统计 NDVI 范围：
    ndvi_values = ndvi_col.select('NDVI').reduce(ee.Reducer.percentile([5, 95]))  # 使用5%和95%分位数
    ndvi_min = ndvi_values.select('NDVI_p5')
    ndvi_max = ndvi_values.select('NDVI_p95')
    stats = ndvi_values.reduceRegion( #统计NDVI
        reducer=ee.Reducer.mean(),# 对整个区域计算平均值
        geometry=roi,
        scale=REDUCE_SCALE, # 降尺度为1.5km
        maxPixels=1e10,# 最大像元数，避免溢出
        bestEffort=True# 若像元过多自动降低分辨率
    ).getInfo()
    print(f"NDVI最小值(5%): {stats.get('NDVI_p5', 'N/A'):.3f}, 最大值(95%): {stats.get('NDVI_p95', 'N/A'):.3f}")
    
    def compute_fvc(img):
        ndvi = img.select('NDVI')
        # FVC = (NDVI - NDVIsoil) / (NDVIveg - NDVIsoil)
        fvc = ndvi.subtract(ndvi_min).divide(ndvi_max.subtract(ndvi_min)).rename('FVC')
        # 限制FVC在0-1范围内
        fvc = fvc.where(fvc.lt(0), 0).where(fvc.gt(1), 1)
        return img.addBands(fvc)# 将计算结果以 “FVC” 波段添加回影像
    
    return ndvi_col.map(compute_fvc) # 为每个影像添加 NDVI 与 FVC 波段，形成新的集合
```

```python
# 有效像元比例计算 
def compute_valid_pct(img, band='NDVI'):
    try:
        # 获取掩膜
        valid_mask = img.select(band).mask()
        
        # 计算有效像素数量，掩膜值为1的像元数（无效掩膜值为0）
        stats = valid_mask.reduceRegion(
            reducer=ee.Reducer.sum(),# sum()求掩膜值之和，有效像元数量
            geometry=roi,
            scale=REDUCE_SCALE,
            maxPixels=1e9,
            bestEffort=True)
        
        valid_count = stats.get(band) # 有效像元数量
        
        # 计算总像素数量
        total_stats = img.select(band).reduceRegion(
            reducer=ee.Reducer.count(),# count()统计所有像元个数
            geometry=roi, 
            scale=REDUCE_SCALE,
            maxPixels=1e9,
            bestEffort=True
        )
        
        total_count = total_stats.get(band)
        
        # 获取结果并计算比例
        valid_val = valid_count.getInfo()
        total_val = total_count.getInfo()
        if valid_val is not None and total_val is not None and total_val > 0:
            pct = (valid_val / total_val) * 100
            return pct
        else:
            return 0.0
            
    except Exception as e:
        print(f"计算有效像元比例时出错: {e}")
        return 0.0
```

```python
#年度和季节合成
annual_list, seasonal_list = [], [] # 用于存储年度与季节影像对象
for year in range(START_YEAR, END_YEAR + 1):
    
    print(f"\n处理年份: {year}")
    col = get_landsat_collection(year)# 获取年度影像集合
    col_size = col.size().getInfo()# 获取影像数量（GEE端到Python端）
    print(f"  {year}年原始影像数量: {col_size}")
    if col_size == 0:
        print(f"  {year}: 无可用影像，跳过")
        continue  
             
    col_with_indices = add_ndvi_fvc(col) # 在集合内添加NDVI、FVC波段
    
    #年度合成
    if col_with_indices.size().getInfo() > 0:
        # 均值合成
        annual_img = col_with_indices.mean().clip(roi).set('year', year)
        # 计算有效像元比例
        ndvi_pct = compute_valid_pct(annual_img, 'NDVI')
        fvc_pct = compute_valid_pct(annual_img, 'FVC')
        # 将有效率信息写入影像属性中
        annual_img = annual_img.set('ndvi_valid_pct', ndvi_pct).set('fvc_valid_pct', fvc_pct)
        annual_list.append(annual_img)# 添加至年度列表
        print(f"  {year}年度: NDVI有效像元{ndvi_pct:.2f}%, FVC有效像元{fvc_pct:.2f}%")
    else:
        print(f"  {year}: 计算指数后无可用影像")

    #季节合成
    for s_name, months in SEASONS.items():
        if s_name == 'winter':
            # 冬季跨年处理
            start = f'{year}-12-01'
            end = f'{year+1}-03-01'
        else:
            start = f'{year}-{months[0]:02d}-01'
            end = f'{year}-{months[-1]+1:02d}-01'
        
        col_s = col_with_indices.filterDate(start, end)# 筛选季节影像
        col_s_size = col_s.size().getInfo()
        
        if col_s_size > 0:
            seasonal_img = col_s.mean().clip(roi).set('year', year).set('season', s_name)
            
            ndvi_pct = compute_valid_pct(seasonal_img, 'NDVI')
            fvc_pct = compute_valid_pct(seasonal_img, 'FVC')
            
            seasonal_img = seasonal_img.set('ndvi_valid_pct', ndvi_pct).set('fvc_valid_pct', fvc_pct)
            seasonal_list.append(seasonal_img)
            print(f"  {year} {s_name}: {col_s_size}景影像, NDVI有效像元{ndvi_pct:.2f}%")
        else:
            print(f"  {year} {s_name}: 无可用影像")

# 转为 ImageCollection
annual_ic = ee.ImageCollection(annual_list)
seasonal_ic = ee.ImageCollection(seasonal_list)

print(f"\n处理完成！年度影像: {annual_ic.size().getInfo()}景, 季节影像: {seasonal_ic.size().getInfo()}景")
```

```python
# 初始化地图
m = geemap.Map(center=[33.95, 113.44], zoom=6, height=300)#河南，缩放因子为6，高度300像素
m.addLayer(roi, {'color': 'red', 'fillColor': '00000000'}, 'Henan boundary')#添加河南边界图层到地图
# 定义可视化参数
vis_ndvi = {
    'min': -0.2, 
    'max': 0.8, 
    'palette': ['blue', 'white', 'yellow', 'green']}
vis_fvc = {
    'min': 0, 
    'max': 1, 
    'palette': ['brown', 'yellow', 'green']}

# 构建图层集合（不提前加载）
annual_layers = {}#年
season_layers = {}#季节
# 遍历年度列表
for i, img in enumerate(annual_list):
    try:
        y = int(img.get('year').getInfo())# 从影像属性中读取年份
        ndvi_band = img.select('NDVI')
        annual_layers[f'NDVI {y}'] = ndvi_band
        # 检查是否有FVC波段
        band_names = img.bandNames().getInfo()
        if 'FVC' in band_names:
            fvc_band = img.select('FVC')
            annual_layers[f'FVC {y}'] = fvc_band
    except Exception as e:
        print(f"  处理年度影像{i}时出错: {e}")
#季节
for i, img in enumerate(seasonal_list):
    try:
        y = int(img.get('year').getInfo())
        s = img.get('season').getInfo()
        ndvi_band = img.select('NDVI')
        season_layers[f'NDVI {s} {y}'] = ndvi_band
        # 检查是否有FVC波段
        band_names = img.bandNames().getInfo()
        if 'FVC' in band_names:
            fvc_band = img.select('FVC')
            season_layers[f'FVC {s} {y}'] = fvc_band
    except Exception as e:
        print(f"  处理季度影像{i}时出错: {e}")

#动态添加图层函数，显示指定年份的年度和季度图层
def show_year_quarters(y, show_annual=False, show_seasonal=True):
    # 清除除边界外的所有图层
    layers_to_remove = []
    for layer_name in m.ee_layer_names:
        if layer_name != 'Henan boundary':
            layers_to_remove.append(layer_name)
    
    for layer_name in layers_to_remove:
        m.remove_layer(layer_name)
    
    added_layers = []# 记录已加载的图层
    
    # 添加年度图层
    if show_annual:
        ndvi_key = f'NDVI {y}'
        fvc_key = f'FVC {y}'
        
        if ndvi_key in annual_layers:
            m.addLayer(annual_layers[ndvi_key], vis_ndvi, ndvi_key)
            added_layers.append(ndvi_key)
        
        if fvc_key in annual_layers:
            m.addLayer(annual_layers[fvc_key], vis_fvc, fvc_key)
            added_layers.append(fvc_key)
    
    # 添加季度图层
    if show_seasonal:
        season_keys = [key for key in season_layers.keys() if f" {y}" in key]
        for key in season_keys:
            if 'NDVI' in key:
                m.addLayer(season_layers[key], vis_ndvi, key)
            else:
                m.addLayer(season_layers[key], vis_fvc, key)
            added_layers.append(key)
    
    if not added_layers:
        print(f"  未找到{y}年的可用图层")
# 交互控件
# 获取可用年份
available_years = list(range(START_YEAR, END_YEAR + 1))
# 年份下拉菜单
year_dropdown = Dropdown(
    options=available_years,
    value=END_YEAR,
    description='年份:',
    style={'description_width': 'initial'})
# 显示年度NDVI/FVC开关
annual_checkbox = Checkbox(
    value=True,
    description='显示年度 NDVI/FVC',
    indent=False)
# 显示季度NDVI/FVC开关
seasonal_checkbox = Checkbox(
    value=True,
    description='显示季度 NDVI/FVC',
    indent=False)
# 图层类型选择框
layer_type_dropdown = Dropdown(
    options=['NDVI', 'FVC', '两者都显示'],
    value='两者都显示',
    description='图层类型:',
    style={'description_width': 'initial'})

# 定义更新地图函数，根据交互控件变化动态刷新
def update_visualization(change=None):
    selected_year = year_dropdown.value
    show_annual = annual_checkbox.value
    show_seasonal = seasonal_checkbox.value
    layer_type = layer_type_dropdown.value
    
    # 根据选择的图层类型更新可视化参数
    global vis_ndvi, vis_fvc
    
    if layer_type == 'NDVI':
        # 只显示NDVI，隐藏FVC
        show_year_quarters(selected_year, show_annual, show_seasonal)
        # 隐藏FVC图层
        for layer_name in m.ee_layer_names:
            if 'FVC' in layer_name and layer_name != 'Henan boundary':
                m.remove_layer(layer_name)
    elif layer_type == 'FVC':
        # 只显示FVC，隐藏NDVI
        show_year_quarters(selected_year, show_annual, show_seasonal)
        # 隐藏NDVI图层
        for layer_name in m.ee_layer_names:
            if 'NDVI' in layer_name and layer_name != 'Henan boundary':
                m.remove_layer(layer_name)
    else:
        # 显示两者
        show_year_quarters(selected_year, show_annual, show_seasonal)

# 绑定事件
year_dropdown.observe(update_visualization, names='value')
annual_checkbox.observe(update_visualization, names='value')
seasonal_checkbox.observe(update_visualization, names='value')
layer_type_dropdown.observe(update_visualization, names='value')
#  添加图例和说明
# 添加NDVI图例
m.add_colorbar(
    vis_ndvi,
    label="NDVI",
    layer_name="NDVI Colorbar",
    position="bottomright")
# 添加FVC图例  
m.add_colorbar(
    vis_fvc,
    label="FVC",
    layer_name="FVC Colorbar",
    position="bottomright")

# 显示控件与地图
# 创建控制面板
control_panel = VBox([
    HBox([year_dropdown, layer_type_dropdown]),
    HBox([annual_checkbox, seasonal_checkbox])])
# 显示控制面板和地图
display(control_panel)
display(m)
update_visualization()# 默认显示最后一年
m.centerObject(henan, 6) #河南为地图中心
```

```python
#合并十年相同季节
def create_seasonal_composites():
    # 按季节分组
    seasonal_groups = {'spring': [], 'summer': [], 'autumn': [],'winter': []}
    #遍历季节影像列表 seasonal_list
    for img in seasonal_list:
        try: # 从影像属性中读取季节和年份信息
            season = img.get('season').getInfo()
            year = img.get('year').getInfo()
            # 如果该影像的季节属于四季之一，则将其添加到对应分组
            if season in seasonal_groups:
                seasonal_groups[season].append({
                    'image': img,
                    'year': year})
        except Exception as e:
            print(f"处理影像时出错: {e}")
            continue
    
    # 创建季节合成
    seasonal_composites = {}
    
    for season, images in seasonal_groups.items():
        if len(images) > 0:
            print(f"处理 {season} 季节合成 ({len(images)}个年份)...")
            
            # 提取同季节的所有影像
            image_list = [item['image'] for item in images]
            years_list = [item['year'] for item in images]
            
            # 创建影像集合
            season_collection = ee.ImageCollection(image_list)
            
            # 计算中值合成（减少异常值影响）
            ndvi_composite = season_collection.select('NDVI').median().rename(f'NDVI_{season}')
            seasonal_composites[f'NDVI_{season}'] = ndvi_composite
            
            try:
                fvc_composite = season_collection.select('FVC').median().rename(f'FVC_{season}')
                seasonal_composites[f'FVC_{season}'] = fvc_composite
            except:
                pass
            
            # 设置属性
            for key in [f'NDVI_{season}', f'FVC_{season}']:
                if key in seasonal_composites:
                    seasonal_composites[key] = seasonal_composites[key].set({
                        'season': season,
                        'composite_type': 'multi_year_median',# 合成类型：多年中值
                        'years_used': years_list,# 使用的年份列表
                        'year_count': len(years_list),# 使用的年份列表
                        'year_range': f"{min(years_list)}-{max(years_list)}"})# 年份范围字符串
    
    print(f"季节合成完成: 共创建{len(seasonal_composites)}个合成图层")
    return seasonal_composites

# 创建季节合成
seasonal_composites = create_seasonal_composites()

# 将季节合成图添加到图层集合
for key, composite in seasonal_composites.items():
    if 'NDVI' in key:
        season = key.split('_')[1]
        annual_layers[f'{key}_composite'] = composite
    elif 'FVC' in key:
        season = key.split('_')[1]
        annual_layers[f'{key}_composite'] = composite

# 添加季节合成显示功能
composite_button = widgets.Button(
    description='显示十年季节合成',
    button_style='success')

def show_seasonal_composites(b):
    # 清除所有图层（除边界外）
    for layer_name in list(m.ee_layer_names):
        if layer_name != 'Henan boundary':
            m.remove_layer(layer_name)
    
    added_layers = []
    
    # 添加NDVI季节合成
    for key in ['NDVI_spring_composite', 'NDVI_summer_composite', 'NDVI_autumn_composite', 'NDVI_winter_composite']:
        if key in annual_layers:
            composite_img = annual_layers[key]
            year_range = composite_img.get('year_range').getInfo() if composite_img.get('year_range') else "2013-2023"
            layer_name = f"{key.replace('_composite', '')} ({year_range}合成)"
            m.addLayer(composite_img, vis_ndvi, layer_name)
            added_layers.append(layer_name)
    
    # 添加FVC季节合成
    for key in ['FVC_spring_composite', 'FVC_summer_composite', 'FVC_autumn_composite', 'FVC_winter_composite']:
        if key in annual_layers:
            composite_img = annual_layers[key]
            year_range = composite_img.get('year_range').getInfo() if composite_img.get('year_range') else "2013-2023"
            layer_name = f"{key.replace('_composite', '')} ({year_range}合成)"
            m.addLayer(composite_img, vis_fvc, layer_name)
            added_layers.append(layer_name)
    
    if not added_layers:
        print("未找到季节合成图层")
    else:
        print(f"成功加载 {len(added_layers)} 个季节合成图层")

composite_button.on_click(show_seasonal_composites)
display(composite_button)
```

```python
# 年度变化折线图
# 计算并导出年度/季度统计到 Pandas，使用 Earth Engine 的 reduceRegion(s) 计算研究区内每年 NDVI / FVC 的平均值，
#并将结果导出为 Pandas DataFrame，方便本地绘图或分析。
def compute_annual_mean_df(annual_ic, region, scale=REDUCE_SCALE, bands=['NDVI','FVC']):
    # 为每幅图像把 year 属性作为属性并用 reduceRegion 计算 mean
    records = []

    # 使用 server-side mapping 构造一个FeatureCollection of (year, mean_band1, mean_band2)
    def img_to_feat(img):
        year = img.get('year') # 获取影像的年份属性
        mean_dict = img.reduceRegion( # 对影像在指定 region 上计算波段平均值
            reducer=ee.Reducer.mean(),
            geometry=region, 
            scale=scale,
            bestEffort=True)
        feat = ee.Feature(None, mean_dict).set('year', year) # 将结果转换为 Feature，对应每年
        return feat
        
    feats = annual_ic.map(img_to_feat).filter(ee.Filter.notNull(['year']))  # 去掉空值
    # 转为小表并拉到本地
    feats_info = feats.getInfo()
    
    for f in feats_info['features']:
        props = f['properties']
        row = {'year': int(props.get('year'))}
        for b in bands:
            if props.get(b) is not None:
                row[b] = props.get(b)
            else:
                row[b] = None
        records.append(row)
    # 转换为 Pandas DataFrame 并按年份排序
    df = pd.DataFrame(records)
    df = df.sort_values('year').reset_index(drop=True)
    return df

# 绘图函数 
def plot_annual_timeseries(df):
    plt.figure(figsize=(9,4))# 设置图像尺寸
    if 'NDVI' in df.columns:# 如果表格中包含 FVC 列，则绘制 FVC 折线
        plt.plot(df['year'], df['NDVI'], marker='o', label='NDVI')
    if 'FVC' in df.columns:
        plt.plot(df['year'], df['FVC'], marker='o', label='FVC')
    # 添加坐标轴标签、标题、网格线和图例
    plt.xlabel('Year'); 
    plt.grid(True); 
    plt.title('Annual NDVI / FVC ({}-{})'.format(START_YEAR, END_YEAR))
    plt.legend();
    plt.show()

# 计算年度均值表并绘图
df_annual = compute_annual_mean_df(annual_ic, henan, scale=REDUCE_SCALE)
print(df_annual)
plot_annual_timeseries(df_annual)
```

```python
# 季度箱线图，对每个季节性影像随机采样若干点，提取 FVC 值，用于绘制季度箱线图。
def compute_quarterly_samples(seasonal_ic, region, sample_points=1000, scale=REDUCE_SCALE):
    recs = []
    # 为每季随机生成 N 点并 sample
    seasons = seasonal_ic.aggregate_array('season').getInfo()
    # 由于 seasonal_ic 可能包含重复season/year，我们把unique image objects 枚举
    imgs = seasonal_ic.toList(seasonal_ic.size())# 将 ImageCollection 转换为 Python 可操作的列表对象
    n = seasonal_ic.size().getInfo()# 获取影像总数
    # 遍历每个季节影像
    for i in range(n):
        img = ee.Image(imgs.get(i))
        year = int(img.get('year').getInfo())
        season = img.get('season').getInfo()
        if 'FVC' in img.bandNames().getInfo():# 只处理包含 FVC 波段的影像
            # 在 region 内随机生成 sample_points 个点
            pts = ee.FeatureCollection.randomPoints(region, sample_points, 
                                                    seed=year*10 + hash(season)%1000)
            # 在 region 内随机生成 sample_points 个点
            samp = img.select('FVC').sampleRegions(collection=pts, scale=scale)
            samp_info = samp.getInfo()# 将结果拉回本地
            for f in samp_info['features']:
                val = f['properties'].get('FVC')
                if val is not None:# 存储为一条记录，包含年份、季节、FVC值
                    recs.append({'year': year, 'season': season, 'fvc': val})
    dfq = pd.DataFrame(recs)# 将采样结果转换为 Pandas DataFrame
    return dfq
    
# 绘制季度 FVC 箱线图
def plot_quarterly_boxline(dfq):
    if dfq.empty:
        print("无季度样本数据可绘制")
        return
    # 箱线图：每个 (year-season) 为一组
    dfq['label'] = dfq['year'].astype(str) + '-' + dfq['season']
    labels = sorted(dfq['label'].unique())# 获取所有年份-季节组合并排序
    # 按标签顺序提取各组数据，用于箱线图输入
    data = [dfq[dfq['label']==lbl]['fvc'].values for lbl in labels]
    #绘图
    plt.figure(figsize=(12,5))
    plt.boxplot(data, labels=labels, showfliers=False)
    plt.xticks(rotation=90);
    plt.ylabel('FVC');
    plt.title('Quarterly FVC distributions')
    plt.tight_layout(); 
    plt.show()

# 计算季度箱线数据并绘图
df_quarter = compute_quarterly_samples(seasonal_ic, henan, sample_points=500)
plot_quarterly_boxline(df_quarter)
```

```python
#特定事件对比分析，
#计算 Landsat 影像 NDVI 与 FVC 变化，并与往年同期均值对比。
def event_before_after(event_date, window_days=90, region=henan, ref_years=5):
    dt = pd.to_datetime(event_date).date()# 将输入字符串转换为日期对象
    year = dt.year# 当前事件年份
     # 计算前后窗口的起止日期
    before_start = (dt - pd.Timedelta(days=window_days)).strftime('%Y-%m-%d')
    before_end   = (dt - pd.Timedelta(days=1)).strftime('%Y-%m-%d')
    after_start  = (dt + pd.Timedelta(days=1)).strftime('%Y-%m-%d')
    after_end    = (dt + pd.Timedelta(days=window_days)).strftime('%Y-%m-%d')

    # 标准化波段定义 
    LC8_BANDS = ['SR_B2','SR_B3','SR_B4','SR_B5','SR_B6','SR_B7','QA_PIXEL']
    LC7_BANDS = ['SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7','QA_PIXEL']
    STD_NAMES = ['blue','green','red','nir','swir1','swir2','QA_PIXEL']

    # 加载并预处理 Landsat 影像集合
    def prepare_landsat(collection_id, band_list):
        return (ee.ImageCollection(collection_id)
                .filterBounds(region)
                .map(apply_scale_factors)# 应用比例系数，DN值转反射率
                .select(band_list, STD_NAMES)# 波段重命名
                .map(mask_clouds))

    l8 = prepare_landsat('LANDSAT/LC08/C02/T1_L2', LC8_BANDS)
    l7 = prepare_landsat('LANDSAT/LE07/C02/T1_L2', LC7_BANDS)
    landsat_all = l8.merge(l7)

    def to_ndvi(img): 
        return img.normalizedDifference(['nir','red']).rename('NDVI')

    # 当前事件年份的前后影像 
    col_before = landsat_all.filterDate(before_start, before_end)
    col_after  = landsat_all.filterDate(after_start, after_end)

    if col_before.size().getInfo() == 0 or col_after.size().getInfo() == 0:
        print("当前年份影像不足，无法计算事件前后。")
        return None

    ndvi_before = col_before.map(to_ndvi).mean()
    ndvi_after  = col_after.map(to_ndvi).mean()

    # 往年同期
    ref_start = (dt.replace(year=year - ref_years) - pd.Timedelta(days=window_days)).strftime('%m-%d')
    ref_end   = (dt.replace(year=year - ref_years) + pd.Timedelta(days=window_days)).strftime('%m-%d')
    ref_start_full = f'{year - ref_years}-01-01'
    ref_end_full = f'{year - 1}-12-31'

    # 动态筛选往年同一时段
    ref_ic = landsat_all.filter(ee.Filter.calendarRange(year - ref_years, year - 1, 'year')) \
                        .filter(ee.Filter.calendarRange(int(ref_start[:2]), int(ref_end[:2]), 'month'))

    ndvi_ref = ref_ic.map(to_ndvi).mean()

    #  FVC计算 
    ndvi_min = ee.ImageCollection([ndvi_before, ndvi_after, ndvi_ref]).min()
    ndvi_max = ee.ImageCollection([ndvi_before, ndvi_after, ndvi_ref]).max()

    def ndvi_to_fvc(ndvi):
        return ndvi.subtract(ndvi_min).divide(ndvi_max.subtract(ndvi_min)).clamp(0,1)

    fvc_before = ndvi_to_fvc(ndvi_before).rename('FVC_before')
    fvc_after  = ndvi_to_fvc(ndvi_after).rename('FVC_after')
    fvc_ref    = ndvi_to_fvc(ndvi_ref).rename('FVC_ref')
    fvc_diff   = fvc_after.subtract(fvc_ref).rename('FVC_diff')

    # 地图显示 
    clear_layers()
    m.addLayer(fvc_ref.clip(region), vis_fvc, f'往年同期 FVC({year-ref_years}-{year-1})')
    m.addLayer(fvc_before.clip(region), vis_fvc, '事件前 FVC')
    m.addLayer(fvc_after.clip(region), vis_fvc, '事件后 FVC')
    m.addLayer(fvc_diff.clip(region), {'min':-0.5,'max':0.5,'palette':
                                       ['red','white','green']}, 
               '事件后 - 往年同期差值')

    print(f"已生成 {year} 年事件 FVC 前后 + 往年同期对比 (±{window_days} 天, 参考{ref_years}年均值)")
    return fvc_before, fvc_after, fvc_ref, fvc_diff
```

```python
# 地貌单元分区统计 FVC：平原、丘陵、山地
def zonal_stats_by_geomorpho(ic, geomorpho_fc, stat='mean', scale=REDUCE_SCALE, band='FVC'):

    # 统计方法选择
    reducer_dict = {
        'mean': ee.Reducer.mean(),
        'median': ee.Reducer.median(),
        'max': ee.Reducer.max(),
        'min': ee.Reducer.min()}
    reducer = reducer_dict.get(stat, ee.Reducer.mean())

    rows = []
    years = ic.aggregate_array('year').getInfo()
    unique_years = sorted(list(set([int(y) for y in years if y is not None])))

    for y in unique_years:
        img = ic.filter(ee.Filter.eq('year', y)).first().select(band)
        if img is None:
            continue
        
        reductions = img.reduceRegions(
            collection=geomorpho_fc,
            reducer=reducer,
            scale=scale
        )
        info = reductions.getInfo()
        for feat in info['features']:
            props = feat['properties']
            zone = props.get('class') or props.get('type') or props.get('NAME') or props.get('system:index')
            rows.append({'year': y, 'zone': zone, f'{stat}_{band.lower()}': props.get(stat)})
    
    df = pd.DataFrame(rows)
    pivot = df.pivot_table(index='year', columns='zone', values=f'{stat}_{band.lower()}')
    return pivot
# 简易地貌分区示例（用SRTM高程自动划分平原/丘陵/山地）
srtm = ee.Image('USGS/SRTMGL1_003').clip(henan)
elev = srtm.select('elevation')

# 平原：<200m，丘陵：200–500m，山地：>500m
plain = elev.lt(200).selfMask().rename('plain')
hill = elev.gte(200).And(elev.lt(500)).selfMask().rename('hill')
mountain = elev.gte(500).selfMask().rename('mountain')

# 转换为 FeatureCollection
plain_fc = plain.reduceToVectors(scale=5000, geometry=henan, labelProperty='class', geometryType='polygon').map(lambda f: f.set('class', '平原'))
hill_fc = hill.reduceToVectors(scale=5000, geometry=henan, labelProperty='class', geometryType='polygon').map(lambda f: f.set('class', '丘陵'))
mountain_fc = mountain.reduceToVectors(scale=5000, geometry=henan, labelProperty='class', geometryType='polygon').map(lambda f: f.set('class', '山地'))

geomorpho_fc = plain_fc.merge(hill_fc).merge(mountain_fc)

pivot_geomorpho = zonal_stats_by_geomorpho(annual_ic, geomorpho_fc, stat='mean', scale=REDUCE_SCALE, band='FVC')
print(pivot_geomorpho)

# 添加地貌层到地图
vis_elev = {
    'min': 50,     # 河南最低约50米
    'max': 2200,   # 太行山、伏牛山最高约2000+米
    'palette': [
        '#006400',  # 深绿：低地
        '#32CD32',  # 亮绿：平原
        '#FFD700',  # 金黄：丘陵
        '#FF8C00',  # 橙色：高丘
        '#FF4500',  # 红色：山地
        '#B22222',  # 深红：高山
        '#800080'   # 紫色：极高山
    ]
}

m.addLayer(elev, vis_elev, '河南省高程图')

# 地貌分区边界
m.addLayer(
    geomorpho_fc.style(**{
        'color': 'magenta',      
        'fillColor': '00000000', # 透明填充
        'width': 2}),{},'地貌分区边界')
m.centerObject(henan, 6) 
```

```python
# 各地貌区不同季节FVC
def seasonal_zonal_stats(seasonal_composites, geomorpho_fc, scale=500):
    results = []
    # 遍历春夏秋冬四个季节
    for season in ['spring', 'summer', 'autumn', 'winter']:
        key = f'FVC_{season}'
        if key not in seasonal_composites:
            print(f"跳过 {season}，未找到合成影像")
            continue
        
        img = seasonal_composites[key]
        print(f"正在计算 {season} 季节...")
        
        # 对每个地貌区计算FVC平均值
        reduction = img.reduceRegions(
            collection=geomorpho_fc,
            reducer=ee.Reducer.mean(),
            scale=scale)
        
        info = reduction.getInfo()
        # 提取地貌区名称和统计值
        for feat in info['features']:
            props = feat['properties']
            geom_name = (
                props.get('zone') or
                props.get('Zone') or
                props.get('Name') or
                props.get('name') or
                props.get('Type') or
                props.get('地貌类型') or
                props.get('class') or 
                '未知地貌')
            mean_val = props.get('mean', None)
            if mean_val is not None:
                results.append({
                    'season': season,
                    'zone': geom_name,
                    'FVC': mean_val})
    
    # 转换为DataFrame表格
    df = pd.DataFrame(results)
    if df.empty:
        print("没有提取到任何统计结果，请检查地貌区字段名或数据范围。")
        return df, None
    
    # 创建透视表，行是地貌区，列是季节
    pivot = df.pivot_table(index='zone', columns='season', values='FVC', aggfunc='mean')
    return df, pivot

# 执行季节分区统计并显示结果
df, seasonal_pivot = seasonal_zonal_stats(seasonal_composites, geomorpho_fc, scale=REDUCE_SCALE)
display(df.head())
display(seasonal_pivot)
```

```python
# 精度检验
years = list(range(2015, 2025))  # 2015到2024
N_SAMPLES = 2000

#逐年验证函数 
def validate_fvc_year(year, roi, n_samples=2000):
    print(f"开始验证 {year} 年...")
    
    try:
        #获取 Landsat 年度 FVC
        landsat_img = ee.ImageCollection(annual_ic).filter(ee.Filter.eq('year', year)).first()
        landsat_fvc = landsat_img.select('FVC')
        
        if landsat_fvc is None:
            print(f"没有找到 {year} 年的 Landsat FVC 图像")
            return None

        # 获取 MODIS FVC 产品
        modis_ic = ee.ImageCollection('MODIS/061/MOD15A2H') .filterDate(f'{year}-01-01', f'{year}-12-31') .select('Fpar_500m')
        modis_fvc = modis_ic.mean().clip(roi).divide(100).rename('FVC_ref')

        #Landsat FVC 重采样到 500 m
        landsat_fvc_500m = landsat_fvc .setDefaultProjection('EPSG:4326', None, 1500) .reduceResolution(ee.Reducer.mean(), maxPixels=4096) .reproject('EPSG:4326', None, 500)

        # 构建配对影像并采样
        pair = landsat_fvc_500m.addBands(modis_fvc).rename(['FVC', 'FVC_ref'])
        sample = pair.sample(region=roi, scale=500, numPixels=n_samples, seed=1)

        count = sample.size().getInfo()
        print(f"{year}年抽取样本数量: {count}")

        if count < 100:  # 样本太少
            print(f"{year}年样本数量不足")
            return None

        # 均值
        mean_fvc = ee.Number(sample.reduceColumns(reducer=ee.Reducer.mean(), 
                                                  selectors=['FVC']).get('mean'))
        mean_ref = ee.Number(sample.reduceColumns(reducer=ee.Reducer.mean(), 
                                                  selectors=['FVC_ref']).get('mean'))

        # 3) 皮尔逊相关系数（直接用内置 reducer）
        corr = ee.Number(sample.reduceColumns(ee.Reducer.pearsonsCorrelation(), 
                                              ['FVC', 'FVC_ref']).get('correlation'))

        # 4) RMSE 与 MAE
        d2_fc = sample.map(lambda f: f.set('d2', ee.Number(f.get('FVC'))
            .subtract(f.get('FVC_ref')).pow(2)))
        rmse = ee.Number(d2_fc.aggregate_mean('d2')).sqrt()

        ad_fc = sample.map(lambda f: f.set('ad', ee.Number(f.get('FVC'))
            .subtract(f.get('FVC_ref')).abs()))
        mae = ee.Number(ad_fc.aggregate_mean('ad'))

        # Bias 与 R^2
        bias = mean_fvc.subtract(mean_ref)
        r2 = corr.pow(2)

        # 打包结果
        result = ee.Dictionary({
            'n': count,
            'corr': corr,
            'r2': r2,
            'rmse': rmse,
            'mae': mae,
            'bias': bias,
            'mean_fvc': mean_fvc,
            'mean_ref': mean_ref})

        result_info = result.getInfo()
        print(f"{year}年验证完成: R²={result_info['r2']:.3f}, RMSE={result_info['rmse']:.3f}")
        
        return {'year': year, **result_info}
        
    except Exception as e:
        print(f"{year}年验证失败: {e}")
        return None

# 执行逐年验证
print("开始2015-2024年FVC逐年验证...")
all_results = []

for year in years:
    result = validate_fvc_year(year, roi, N_SAMPLES)
    if result is not None:
        all_results.append(result)
    
    # 添加延迟避免GEE请求限制
    time.sleep(1)

print(f"\n验证完成: {len(all_results)}/{len(years)} 年数据成功验证")

# 结果整理和分析
if all_results:
    # 转换为DataFrame便于分析
    df_results = pd.DataFrame(all_results)
    df_results = df_results.sort_values('year').reset_index(drop=True)
    
    #统计摘要 
    print("\n十年验证统计摘要:")
    print("=" * 50)
    print(f"平均 R²: {df_results['r2'].mean():.4f} ± {df_results['r2'].std():.4f}")
    print(f"平均 RMSE: {df_results['rmse'].mean():.4f} ± {df_results['rmse'].std():.4f}")
    print(f"平均偏差: {df_results['bias'].mean():.4f} ± {df_results['bias'].std():.4f}")
    print(f"平均相关系数: {df_results['corr'].mean():.4f} ± {df_results['corr'].std():.4f}")
    print(f"平均 MAE: {df_results['mae'].mean():.4f} ± {df_results['mae'].std():.4f}")
    
    # 最佳和最差年份
    best_year_r2 = df_results.loc[df_results['r2'].idxmax()]
    worst_year_r2 = df_results.loc[df_results['r2'].idxmin()]
    
    print(f"\n最佳年份 (R²): {best_year_r2['year']} (R² = {best_year_r2['r2']:.4f})")
    print(f"最差年份 (R²): {worst_year_r2['year']} (R² = {worst_year_r2['r2']:.4f})")
    
    # 趋势分析
    r2_trend = df_results['r2'].corr(pd.Series(df_results['year']))
    rmse_trend = df_results['rmse'].corr(pd.Series(df_results['year']))
    
    print(f"\n趋势分析:")
    print(f"R² 年度趋势相关系数: {r2_trend:.4f}")
    print(f"RMSE 年度趋势相关系数: {rmse_trend:.4f}")
    
    if r2_trend > 0.1:
        print("R² 呈上升趋势")
    elif r2_trend < -0.1:
        print("R² 呈下降趋势")
    else:
        print("R² 趋势平稳")
    
    # 详细结果表格
    print("\n" + "=" * 80)
    print("年度详细验证结果:")
    print("=" * 80)
    display_columns = ['year', 'r2', 'rmse', 'bias', 'corr', 'mae', 'n']
    print(df_results[display_columns].round(4).to_string(index=False))
    
else:
    print("没有成功验证任何年份的数据")
```

```python
# 变化趋势分析
# 参数设置 
SLOPE_THRESHOLD = 0.0005  # 斜率阈值
Z_THRESHOLD = 1.96       # Z值阈值（95%置信水平）

# 可视化参数
PALETTE_5 = ['#a50026', '#f46d43', '#ffffbf', '#a6d96a', '#1a9850']
NAMES_5 = ['明显退化', '轻微退化', '稳定不变', '轻微改善', '明显改善']

# 选择FVC波段并确保数据类型一致
fvc_ic = annual_ic.select('FVC').map(lambda img: img.toFloat())

# 线性趋势计算 
def add_xy_variables(img):
    year = ee.Number(img.get('year'))
    # 确保年份波段使用一致的数据类型
    x_band = ee.Image.constant(year).toInt().rename('x')
    y_band = img.select('FVC').rename('y')
    return y_band.addBands(x_band)

# 创建只包含x和y波段的ImageCollection
xy_ic = fvc_ic.map(add_xy_variables)

# 确保只有2个波段：x和y
xy_ic = xy_ic.select(['x', 'y'])

# 检查波段数量
print(f"xy_ic波段数量: {xy_ic.first().bandNames().getInfo()}")

# 计算线性回归
fit = xy_ic.reduce(ee.Reducer.linearFit())
slope = fit.select('scale').rename('slope').clip(roi)
intercept = fit.select('offset').rename('intercept').clip(roi)

print("线性趋势计算完成")

#Mann-Kendall趋势检验 
def mann_kendall(ic):
    n = ee.Number(ic.size())
    img_list = ic.toList(ic.size())
    
    def compute_s(i):
        i = ee.Number(i)
        img_i = ee.Image(img_list.get(i)).select('FVC')
        
        def compare_with_j(j):
            j = ee.Number(j)
            img_j = ee.Image(img_list.get(j)).select('FVC')
            # 使用gt和lt方法计算符号差异
            # 如果img_i > img_j，返回1；如果img_i < img_j，返回-1；否则返回0
            return img_i.gt(img_j).subtract(img_i.lt(img_j))
        
        previous_indices = ee.List.sequence(0, i.subtract(1))
        comparisons = ee.ImageCollection(previous_indices.map(compare_with_j))
        return comparisons.sum()
    
    indices = ee.List.sequence(1, n.subtract(1))
    S = ee.ImageCollection(indices.map(compute_s)).sum()
    
    # 计算方差
    varS = n.multiply(n.subtract(1)).multiply(n.multiply(2).add(5)).divide(18)
    
    # 计算标准化的Z值
    Z = S.divide(varS.sqrt())
    return Z.rename('Z')

# 使用修复的函数计算Mann-Kendall Z值
z_img = mann_kendall(fvc_ic).clip(roi)
print("Mann-Kendall趋势检验完成")

# 趋势分类 
# 确保所有波段使用一致的数据类型
slope = slope.toFloat()
z_img = z_img.toFloat()

trend = slope.addBands(z_img).clip(roi)

trend_class = trend.expression(
    f"(b('slope') < -{SLOPE_THRESHOLD} && b('Z') < -{Z_THRESHOLD}) ? 1"
    f": (b('slope') < -{SLOPE_THRESHOLD} && b('Z') >= -{Z_THRESHOLD} && b('Z') <= {Z_THRESHOLD}) ? 2"
    f": (b('slope') >= -{SLOPE_THRESHOLD} && b('slope') <= {SLOPE_THRESHOLD}) ? 3"
    f": (b('slope') > {SLOPE_THRESHOLD} && b('Z') >= -{Z_THRESHOLD} && b('Z') <= {Z_THRESHOLD}) ? 4"
    f": (b('slope') > {SLOPE_THRESHOLD} && b('Z') > {Z_THRESHOLD}) ? 5"
    ": 0").rename('FVC_trend_class')

# 确保趋势分类使用一致的数据类型
trend_class = trend_class.toByte()

print("趋势分类完成")
# 添加趋势分类图层（主图层）
trend_vis_params = {
    'min': 1, 
    'max': 5, 
    'palette': PALETTE_5,
    'opacity': 0.8}
m.addLayer(trend_class, trend_vis_params, 'FVC趋势分类')

# 添加斜率图层
slope_vis_params = {
    'min': -0.001, 
    'max': 0.001, 
    'palette': ['red', 'white', 'green']}
m.addLayer(slope, slope_vis_params, 'FVC变化斜率')

# 添加Z值图层
z_vis_params = {
    'min': -2.5, 
    'max': 2.5, 
    'palette': ['blue', 'white', 'red']}
m.addLayer(z_img, z_vis_params, 'Mann-Kendall Z值')
# 添加图例
legend_dict = {
    '明显退化': '#a50026',
    '轻微退化': '#f46d43', 
    '稳定不变': '#ffffbf',
    '轻微改善': '#a6d96a',
    '明显改善': '#1a9850'}
m.add_legend(legend_dict=legend_dict, title='FVC趋势分类')

# 添加图层控制
m.addLayerControl()

print("地图创建完成")

# 面积统计与显示 
def calculate_and_display_stats():
    # 每个像元面积 (km²)
    area_img = ee.Image.pixelArea().divide(1e6)
    
    # 确保数据类型一致
    area_img = area_img.toFloat()
    trend_class_for_stats = trend_class.toInt()
    
    # 按类别统计面积
    area_stats = area_img.addBands(trend_class_for_stats).reduceRegion(
        reducer=ee.Reducer.sum().group(groupField=1, groupName='class'),
        geometry=roi,
        scale=1000,
        maxPixels=1e13,
        bestEffort=True)
    
    # 解析统计结果
    try:
        class_stats = ee.List(area_stats.get('groups')).getInfo()
        total_area = sum([item['sum'] for item in class_stats])
        
        # 显示统计结果
        print("\n" + "="*60)
        print("FVC趋势变化统计分析结果")
        print("="*60)
        print(f"研究区域总面积: {total_area:,.2f} km²")
        print("-"*60)
        print(f"{'趋势类别':<10} | {'面积(km²)':>12} | {'百分比(%)':>10}")
        print("-"*60)
        
        stats_data = []
        for item in class_stats:
            cls = int(item['class'])
            area = item['sum']
            percent = area / total_area * 100
            
            if 1 <= cls <= 5:
                trend_name = NAMES_5[cls-1]
                print(f"{trend_name:<10} | {area:>12,.2f} | {percent:>9.2f}%")
                stats_data.append({
                    'class': cls,
                    'name': trend_name,
                    'area_km2': area,
                    'percentage': percent})
        
        print("-"*60)
        
        # 生成简要分析报告
        if stats_data:
            improvement_area = sum([item['area_km2'] for item in stats_data if item['class'] in [4, 5]])
            degradation_area = sum([item['area_km2'] for item in stats_data if item['class'] in [1, 2]])
            
            improvement_percent = improvement_area / total_area * 100
            degradation_percent = degradation_area / total_area * 100
            
            print(f"植被改善区域: {improvement_area:,.2f} km² ({improvement_percent:.2f}%)")
            print(f"植被退化区域: {degradation_area:,.2f} km² ({degradation_percent:.2f}%)")
            
            if improvement_area > degradation_area:
                print("总体趋势: 改善")
            elif improvement_area < degradation_area:
                print("总体趋势: 退化")
            else:
                print("总体趋势: 稳定")
        
        return stats_data, total_area
        
    except Exception as e:
        print(f"计算面积统计时出错: {e}")
        print("跳过面积统计，继续显示地图...")
        return [], 0

# 计算并显示统计信息
stats_data, total_area = calculate_and_display_stats()

# 创建统计图表 
def create_statistics_chart(stats_data):
    if not stats_data:
        print("没有有效的统计数据可绘图")
        return
    
    # 创建图表
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # 准备数据
    df = pd.DataFrame(stats_data)
    colors = [PALETTE_5[i-1] for i in df['class']]
    
    # 饼图
    wedges, texts, autotexts = ax1.pie(
        df['percentage'], 
        labels=df['name'], 
        autopct='%1.1f%%',
        colors=colors,
        startangle=90)
    
    for autotext in autotexts:
        autotext.set_color('white')
        autotext.set_fontweight('bold')
    
    ax1.set_title('FVC趋势变化分布', fontsize=14, fontweight='bold')
    
    # 柱状图
    bars = ax2.bar(df['name'], df['area_km2'], color=colors)
    ax2.set_title('各趋势类别面积', fontsize=14, fontweight='bold')
    ax2.set_ylabel('面积 (平方千米)')
    ax2.tick_params(axis='x', rotation=45)
    
    # 在柱状图上添加数值标签
    for bar, area in zip(bars, df['area_km2']):
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + max(df['area_km2'])*0.01,
                f'{area:,.0f}', ha='center', va='bottom', fontweight='bold')
    
    plt.tight_layout()
    plt.show()

# 显示统计图表
if stats_data:
    create_statistics_chart(stats_data)

m
```
